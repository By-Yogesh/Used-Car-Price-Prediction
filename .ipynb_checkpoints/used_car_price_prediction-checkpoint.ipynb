```python
# Importing necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

### LOADING AND UNDERSTANDING DATA

```python
# Loading the dataset
ucd = pd.read_csv('used_cars_data.csv')
```

```python
# Basic Data Understanding
print("Dataset Shape:", ucd.shape)
print("\nFirst Few Rows:")
display(ucd.head())
print("\nData Types:")
display(ucd.dtypes)
print("\nDataset Info:")
ucd.info()
```

```python
# DESCRIPTIVE STATISTICS OF THE DATASET
ucd.describe()
```

### DATA CLEANING

```python
# FIXING DATA TYPES AND CORRECTING NAMES OF THE FEATURES
# Drop S.No. as it is not useful for analysis
ucd.drop(columns=['S.No.'], inplace=True)

# Remove 'kmpl' from Mileage column and convert to numeric
ucd['Mileage'] = ucd['Mileage'].str.extract(r'(\d+\.?\d*)').astype(float)
ucd.rename(columns={'Mileage': 'Mileage_Kmpl'}, inplace=True)

# Remove 'CC' from Engine column and convert to numeric
ucd['Engine'] = ucd['Engine'].str.extract(r'(\d+\.?\d*)').astype(float)
ucd.rename(columns={'Engine': 'Engine_CC'}, inplace=True)

# Remove 'bhp' from Power column and convert to numeric
ucd['Power'] = ucd['Power'].str.extract(r'(\d+\.?\d*)').astype(float)
ucd.rename(columns={'Power': 'Power_bhp'}, inplace=True)

# Convert Price from Lakh to actual rupees (multiply by 100,000)
ucd['Price'] = ucd['Price'] * 100000

# Extract Brand from Name
ucd['Brand'] = ucd['Name'].str.split().str[0]

# Reorder columns to place Brand after Name
columns = list(ucd.columns)
columns.remove('Brand')
columns.insert(1, 'Brand')
ucd = ucd[columns]
```

```python
# DISPLAYING THE FIXED DATASET
ucd.head()
```

```python
# Missing Values Analysis
print("Missing Values:")
missing_df = pd.DataFrame({
    'Column': ucd.columns,
    'Missing_Count': ucd.isnull().sum(),
    'Missing_Percentage': (ucd.isnull().sum() / len(ucd) * 100).round(2)
})
print(missing_df[missing_df['Missing_Count'] > 0].sort_values('Missing_Count', ascending=False))
print("\nStatistical Summary:")
print(ucd.describe())
```

```python
# Visualize missing values
plt.figure(figsize=(10, 5))
sns.heatmap(ucd.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Value Heatmap')
plt.show()
```

```python
# HANDLING MISSING VALUES
# Drop rows with missing values in Seats, Power_bhp, Engine_CC, and Mileage_Kmpl
ucd.dropna(subset=['Seats', 'Power_bhp', 'Engine_CC', 'Mileage_Kmpl'], inplace=True)

# Final dataset shape after handling missing values
print(f"Dataset shape after handling missing values: {ucd.shape}")

# Fill Median values Where Mileage is 0
mileage_median = ucd[ucd['Mileage_Kmpl'] > 0]['Mileage_Kmpl'].median()
ucd.loc[ucd['Mileage_Kmpl'] == 0, 'Mileage_Kmpl'] = mileage_median
print(f"Filled {(ucd['Mileage_Kmpl'] == 0).sum()} zero values in Mileage_Kmpl with median: {mileage_median:.2f}")
```

```python
# OUTLIER HANDLING
columns_to_check = ['Year', 'Kilometers_Driven', 'Mileage_Kmpl','Engine_CC', 'Power_bhp']
# Boxplot to spot outliers visually in a 2x3 subplot
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()
for i, col in enumerate(columns_to_check):
    sns.boxplot(x=ucd[col], ax=axes[i])
    axes[i].set_title(f'Outliers - {col}')
plt.tight_layout()
plt.show()
```

```python
# IQR Method
print("=== Outlier Count (IQR Method) ===")
for col in columns_to_check:
    Q1 = ucd[col].quantile(0.25)
    Q3 = ucd[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    count = ((ucd[col] < lower) | (ucd[col] > upper)).sum()
    pct = count / len(ucd) * 100
    print(f"  {col}: {count} outliers ({pct:.2f}%)")

```

```python
# Filter out outliers
ucd = ucd[(ucd[col] >= lower) & (ucd[col] <= upper)]
print("Outliers removed. New shape:", ucd.shape)
```

### UNIVARIATE ANALYSIS

```python
# Numerical Columns Distribution Analysis
numeric_cols = ['Year', 'Kilometers_Driven', 'Mileage_Kmpl', 'Engine_CC', 'Power_bhp', 'Price']
for col in numeric_cols:
    fig, axes = plt.subplots(1, 2, figsize=(15, 4))

    # Histogram
    ucd[col].hist(ax=axes[0], bins=30, color='steelblue', edgecolor='black')
    axes[0].set_title(f'Histogram — {col}')
    axes[0].set_xlabel(col)
    axes[0].set_ylabel('Frequency')

    # KDE
    ucd[col].dropna().plot(kind='kde', ax=axes[1], color='darkorange')
    axes[1].set_title(f'KDE — {col}')
    axes[0].set_xlabel(col)
    axes[0].set_ylabel('Frequency')


    plt.suptitle(col, fontsize=13, fontweight='bold')
    plt.tight_layout()
    plt.show()


    # Stats
    print(f"\n{col}")
    print(f"  Mean   : {ucd[col].mean():.3f}")
    print(f"  Median : {ucd[col].median():.3f}")
    print(f"  Std    : {ucd[col].std():.3f}")
    print(f"  Skew   : {ucd[col].skew():.3f}")
    print(f"  Kurt   : {ucd[col].kurt():.3f}")
    print("-" * 35)
```

```python
# Categorical Columns Analysis
categorical_cols = ucd[['Brand','Location','Fuel_Type','Transmission','Owner_Type','Seats']]
for col in categorical_cols:
    n_unique = ucd[col].nunique()
    print(f"\n{col} — {n_unique} unique values")
    print(ucd[col].value_counts().head(10))

    fig, axes = plt.subplots(1, 2, figsize=(15, 5))

    # Bar chart (top 15)
    ucd[col].value_counts().head(15).plot(
        kind='bar', ax=axes[0], color='steelblue', edgecolor='black'
    )
    axes[0].set_title(f'Bar Chart — {col}')
    axes[0].set_xlabel('')
    axes[0].tick_params(axis='x', rotation=45)
    axes[0].set_ylabel('Count of Sales')
    axes[0].grid(axis='y', linestyle='--', alpha=0.7)

    # Pie chart (top 8)
    ucd[col].value_counts().head(8).plot(
        kind='pie', ax=axes[1], autopct='%1.1f%%', startangle=90, labels=None
    )
    axes[1].set_xlabel('Percentage of Sales')
    axes[1].set_title(f'Pie Chart — {col}')
    axes[1].legend(ucd[col].value_counts().head(8).index, loc='lower left', fontsize=8)


    plt.tight_layout()
    plt.show()
```

### Bivariate Analysis:

```python
# Numerical Columns vs Price (Price on Y-axis)

price_col = ucd['Price']  # ← CHANGE if your column name differs

price_col = 'Price'
other_num_cols = [col for col in numeric_cols if col != price_col]

for col in other_num_cols:
    fig, ax = plt.subplots(figsize=(7, 5))

    ax.scatter(
        ucd[col], ucd[price_col],
        alpha=0.5, color='steelblue',
        edgecolors='white', linewidths=0.3
    )

    # Trendline
    clean = ucd[[col, price_col]].dropna()
    m, b = np.polyfit(clean[col], clean[price_col], 1)
    ax.plot(clean[col], m * clean[col] + b, color='red', linewidth=1.5, label='Trend')

    # Y-axis — price labels
    ax.yaxis.set_major_formatter(
        plt.FuncFormatter(lambda x, _: f'{x:,.0f}')
    )

    # Labels
    ax.set_xlabel(col, fontsize=12)
    ax.set_ylabel('Price', fontsize=12)

    corr = clean[col].corr(clean[price_col])
    ax.set_title(f'{col}  vs  Price  |  r = {corr:.2f}', fontsize=13, fontweight='bold')
    ax.legend()

    plt.tight_layout()
    plt.show()

    print(f"Correlation of {col} with Price: {corr:.3f}")
    print("-" * 40)
```

```python
price_col = 'Price'

for col in categorical_cols:
    clean = ucd[[col, price_col]].dropna()

    # Keep top 10 categories by median price
    top_cats = (
        clean.groupby(col)[price_col]
        .median()
        .sort_values(ascending=False)
        .head(10)
        .index
    )
    clean = clean[clean[col].isin(top_cats)]

    # Order by median price
    order = (
        clean.groupby(col)[price_col]
        .median()
        .sort_values(ascending=False)
        .index
    )

    plt.figure(figsize=(10, 6))

    sns.violinplot(
        data=clean,
        x=col,
        y=price_col,
        order=order,
        palette='muted'
    )

    # Y-axis price labels
    plt.gca().yaxis.set_major_formatter(
        plt.FuncFormatter(lambda x, _: f'{x:,.0f}')
    )

    # Median price annotation on each box
    medians = clean.groupby(col)[price_col].median()
    for j, category in enumerate(order):
        plt.text(
            j, medians[category],
            f'{medians[category]:,.0f}',
            ha='center', va='bottom',
            fontsize=9, color='black', fontweight='bold'
        )

    plt.title(f'{col} vs Price', fontsize=13, fontweight='bold')
    plt.xlabel(col, fontsize=11)
    plt.ylabel('Price', fontsize=11)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    print(f"\nMedian Price by {col}:")
    print(medians.reindex(order).apply(lambda x: f'{x:,.0f}'))
    print("-" * 40)
```

### FEATURE ENGINEERING

```python
# 'Name' column as it's high cardinality and not suitable for one-hot encoding in this context
ucd = ucd.drop('Name', axis=1)

# Car Age — more useful than Year directly
ucd['Car_Age'] = 2024 - ucd['Year']

# Drop Year — no longer needed
ucd.drop(columns=['Year'], inplace=True)

```

```python
ucd.head()
```

```python
# One Hot Encoding
ucd_encoded = pd.get_dummies(ucd, columns=['Brand', 'Location', 'Fuel_Type', 'Transmission', 'Owner_Type'], drop_first=True)

print("Shape after encoding:", ucd.shape)
ucd_encoded.head()
```

```python
ucs_encoded = ucd_encoded.astype(int)
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
```

```python
# Separate features and target
X = ucd_encoded.drop(columns=['Price'])
y = ucd_encoded['Price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print(f"\nShape of X_train: {X_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test: {y_test.shape}")
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Initialize the Linear Regression model
lr = LinearRegression()

# Train the model
lr.fit(X_train_scaled, y_train)

# Predictions
y_pred_lr = lr.predict(X_test_scaled)

# Evaluate the model
MSE_lr = mean_squared_error(y_test, y_pred_lr)
R2_lr = r2_score(y_test, y_pred_lr)
RMSE_lr = np.sqrt(MSE_lr)

print(f"\nLinear Regression Model Performance:")
print(f"Mean Squared Error (MSE): {MSE_lr:.2f}")
print(f"RMSE: {RMSE_lr:.2f}")
print(f"R-squared (R2): {R2_lr:.2f}")
```

```python
# Initialize and train Random Forest
rf = RandomForestRegressor(n_estimators=100, random_state=42)
rf.fit(X_train_scaled, y_train)

# Predictions
y_pred_rf = rf.predict(X_test_scaled)

# Evaluation
mse_rf = mean_squared_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mse_rf)
r2_rf = r2_score(y_test, y_pred_rf)

print(f"Random Forest Performance:")
print(f"RMSE: {rmse_rf:.2f}")
print(f"R2 Score: {r2_rf:.2f}")
```

```python
# Initialize and train Gradient Boosting
gb = GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, max_depth=5, random_state=42)
gb.fit(X_train_scaled, y_train)

# Predictions
y_pred_gb = gb.predict(X_test_scaled)

# Evaluation
mse_gb = mean_squared_error(y_test, y_pred_gb)
rmse_gb = np.sqrt(mse_gb)
r2_gb = r2_score(y_test, y_pred_gb)

print(f"Gradient Boosting Performance:")
print(f"RMSE: {rmse_gb:.2f}")
print(f"R2 Score: {r2_gb:.2f}")
```
