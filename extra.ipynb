```python
# Drop 'Name' column as it's high cardinality and not suitable for one-hot encoding in this context

ucd = ucd.drop(['Name', 'Brand'], axis=1)
```

```python
categorical_cols = [col for col in ['Location', 'Fuel_Type', 'Transmission', 'Owner_Type', 'Seats'] if col in ucd.columns]

# Label encode the higher-cardinality categorical columns
label_encode_cols = [col for col in categorical_cols if ucd[col].nunique() > 10]
for col in label_encode_cols:
    ucd[col + '_encoded'] = LabelEncoder().fit_transform(ucd[col].astype(str))

# One-hot encode the low-cardinality categorical columns
onehot_cols = [col for col in categorical_cols if ucd[col].nunique() <= 10]
ucd = pd.get_dummies(ucd, columns=onehot_cols, drop_first=True)

# Remove the original higher-cardinality columns after label encoding
if label_encode_cols:
    ucd.drop(columns=label_encode_cols, inplace=True)
```
